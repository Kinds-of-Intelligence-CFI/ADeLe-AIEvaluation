```python
"""
K-fold CV (instance-wise) -- ID evaluation
"""
import time
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, accuracy_score, brier_score_loss
from sklearn.calibration import calibration_curve
import matplotlib.pyplot as plt

def calculate_ece(y_true, y_pred, n_bins=10):
    """
    Calculate Expected Calibration Error (ECE).
    ECE is a weighted average of the calibration errors across bins,
    where the weight is the fraction of samples in each bin.
    """
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    bin_edges = np.linspace(0, 1, n_bins + 1)
    bin_indices = np.digitize(y_pred, bin_edges) - 1
    bin_indices = np.minimum(bin_indices, n_bins - 1)
    
    ece = 0
    total_samples = len(y_true)
    
    for i in range(n_bins):
        mask = bin_indices == i
        bin_samples = np.sum(mask)
        if bin_samples > 0:
            acc = np.mean(y_true[mask])
            conf = np.mean(y_pred[mask])
            # Weight by fraction of samples in the bin
            ece += (bin_samples / total_samples) * abs(acc - conf)
    
    return ece

def calculate_balanced_calibration_loss(y_true, y_pred, n_bins=10):
    """
    Calculate balanced calibration loss that treats all bins equally regardless of
    number of instances in each bin.
    """
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    bin_edges = np.linspace(0, 1, n_bins + 1)
    bin_indices = np.digitize(y_pred, bin_edges) - 1
    bin_indices = np.minimum(bin_indices, n_bins - 1)
    
    cal_loss = 0
    n_valid_bins = 0
    
    for i in range(n_bins):
        mask = bin_indices == i
        if np.sum(mask) > 0:
            acc = np.mean(y_true[mask])
            conf = np.mean(y_pred[mask])
            cal_loss += abs(acc - conf)
            print(f"Fold bin {i}: Accuracy = {acc:.3f}, Confidence = {conf:.3f}")
            n_valid_bins += 1
    
    return cal_loss / n_valid_bins if n_valid_bins > 0 else 0

from sklearn.metrics import brier_score_loss

def run_ten_fold_ID_evaluation(X, y, seed=42, model_name='gpt4o'):
    """
    Run Random Forest evaluation using standard 10-fold cross-validation.
    Data is split randomly at the instance/row level.
    """
    kf = KFold(n_splits=10, shuffle=True, random_state=seed)
    
    all_preds = []
    all_preds_binary = []
    all_true = []
    
    fold_training_times = []
    
    for fold_idx, (train_idx, test_idx) in enumerate(kf.split(X)):
        X_train = X.iloc[train_idx]
        X_test = X.iloc[test_idx]
        y_train = y.iloc[train_idx]
        y_test = y.iloc[test_idx]
        
        model = RandomForestClassifier(random_state=seed, 
                                       min_samples_split=50,
                                       criterion='entropy')
        
        # Time the training process for this fold
        start_time = time.time()
        model.fit(X_train, y_train)
        training_time = time.time() - start_time
        fold_training_times.append(training_time)
        print(f"Fold {fold_idx} training time: {training_time:.4f} seconds")
        
        y_test_pred_proba = model.predict_proba(X_test)[:, 1]
        y_test_pred_binary = model.predict(X_test)
        
        all_preds.extend(y_test_pred_proba)
        all_preds_binary.extend(y_test_pred_binary)
        all_true.extend(y_test)
    
    overall_auroc = roc_auc_score(all_true, all_preds)
    overall_accuracy = accuracy_score(all_true, all_preds_binary)
    overall_bal_cal = calculate_balanced_calibration_loss(all_true, all_preds)
    overall_ece = calculate_ece(all_true, all_preds)  # Calculate ECE
    overall_brier = brier_score_loss(all_true, all_preds)  # Calculate Brier score
    
    print(f"\nOverall AUROC: {overall_auroc:.3f}")
    print(f"Overall Accuracy: {overall_accuracy:.3f}")
    print(f"Overall Balanced Calibration Loss: {overall_bal_cal:.3f}")
    print(f"Overall Expected Calibration Error: {overall_ece:.3f}")
    print(f"Overall Brier Score: {overall_brier:.3f}")
    print(f"Average training time per fold: {np.mean(fold_training_times):.4f} seconds")
    
    prob_true, prob_pred = calibration_curve(all_true, all_preds, n_bins=10)
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 12), height_ratios=[2, 1], gridspec_kw={'hspace': 0.1})
    
    bin_edges = np.linspace(0, 1, 11)
    bin_counts, _ = np.histogram(all_preds, bins=bin_edges)
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
    bin_percentages = (bin_counts / len(all_preds)) * 100
    
    ax1.plot(prob_pred, prob_true, marker='o', label='Calibration curve')
    ax1.plot([0, 1], [0, 1], linestyle='--', label='Perfect calibration')
    ax1.set_xlabel('')
    ax1.set_ylabel('True Probability')
    ax1.set_title('Calibration Plot')
    ax1.grid(True)
    ax1.legend()
    
    bars = ax2.bar(bin_centers, bin_percentages, width=0.08, alpha=0.6)
    ax2.set_xlabel('Mean Predicted Probability')
    ax2.set_ylabel('Percentage of Instances (%)')
    ax2.grid(True)
    
    for bar in bars:
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height,
                 f'{height:.1f}%',
                 ha='center', va='bottom')
    
    ax1.set_xlim(-0.05, 1.05)
    ax2.set_xlim(-0.05, 1.05)

    plt.savefig(f'./figures/calibration_ID_subject={model_name}.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    return {
        'overall_auroc': overall_auroc,
        'overall_accuracy': overall_accuracy,
        'overall_balanced_calibration': overall_bal_cal,
        'overall_ece': overall_ece,
        'overall_brier_score': overall_brier,  # Now included in the returned dictionary
        'accumulated_predictions': {
            'y_true': all_true,
            'y_pred_proba': all_preds,
            'y_pred_binary': all_preds_binary
        },
        'fold_training_times': fold_training_times
    }


"""
O.O.D. setup
K-fold CV (Benchmark-wise)
"""
import time
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, accuracy_score
from sklearn.calibration import calibration_curve
import matplotlib.pyplot as plt

def calculate_balanced_calibration_loss(y_true, y_pred, n_bins=10):
    y_true, y_pred = np.array(y_true), np.array(y_pred)
    bin_edges = np.linspace(0, 1, n_bins + 1)
    bin_indices = np.digitize(y_pred, bin_edges) - 1
    bin_indices = np.minimum(bin_indices, n_bins - 1)
    
    cal_loss = 0
    n_valid_bins = 0
    
    for i in range(n_bins):
        mask = bin_indices == i
        if np.sum(mask) > 0:
            acc = np.mean(y_true[mask])
            conf = np.mean(y_pred[mask])
            cal_loss += abs(acc - conf)
            n_valid_bins += 1
    
    return cal_loss / n_valid_bins if n_valid_bins > 0 else 0

def calculate_ece(y_true, y_pred, n_bins=10):
    """
    Calculate Expected Calibration Error (ECE).
    ECE is a weighted average of the calibration errors across bins,
    where the weight is the fraction of samples in each bin.
    """
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    bin_edges = np.linspace(0, 1, n_bins + 1)
    bin_indices = np.digitize(y_pred, bin_edges) - 1
    bin_indices = np.minimum(bin_indices, n_bins - 1)
    
    ece = 0
    total_samples = len(y_true)
    
    for i in range(n_bins):
        mask = bin_indices == i
        bin_samples = np.sum(mask)
        if bin_samples > 0:
            acc = np.mean(y_true[mask])
            conf = np.mean(y_pred[mask])
            # Weight by fraction of samples in the bin
            ece += (bin_samples / total_samples) * abs(acc - conf)
    
    return ece

def run_ten_fold_OOD_evaluation(X, y, processed_df, split_by='benchmark', model_name='gpt4o', seed=42):
    if split_by == 'benchmark':
        unique_values = np.array(processed_df['benchmark'].unique())
    elif split_by == 'task':
        unique_values = np.array(processed_df['task'].unique())
    else:
        raise ValueError("split_by must be either 'benchmark' or 'task'")
    
    np.random.seed(seed)
    np.random.shuffle(unique_values)
    
    kf = KFold(n_splits=10, shuffle=True, random_state=seed)
    value_folds = list(kf.split(unique_values))
    
    all_preds = []
    all_preds_binary = []
    all_true = []
    value_predictions = {}
    
    fold_training_times = []
    
    for fold_idx, (train_val_idx, test_val_idx) in enumerate(value_folds):
        test_values = unique_values[test_val_idx]
        train_values = unique_values[train_val_idx]
        
        test_mask = processed_df[split_by].isin(test_values)
        train_mask = processed_df[split_by].isin(train_values)
        
        X_train = X.loc[train_mask]
        X_test = X.loc[test_mask]
        y_train = y.loc[train_mask]
        y_test = y.loc[test_mask]
        
        model = RandomForestClassifier(random_state=seed, 
                                       min_samples_split=50,
                                       criterion='entropy')
        
        # Time the training process for this fold
        start_time = time.time()
        model.fit(X_train, y_train)
        training_time = time.time() - start_time
        fold_training_times.append(training_time)
        print(f"Fold {fold_idx} training time (OOD): {training_time:.4f} seconds")
        
        y_test_pred_proba = model.predict_proba(X_test)[:, 1]
        y_test_pred_binary = model.predict(X_test)
        
        all_preds.extend(y_test_pred_proba)
        all_preds_binary.extend(y_test_pred_binary)
        all_true.extend(y_test)
        
        # Store predictions per value
        for value in test_values:
            value_mask = processed_df.loc[test_mask, split_by] == value
            if sum(value_mask) > 0:
                if value not in value_predictions:
                    value_predictions[value] = {'true': [], 'pred_proba': [], 'pred_binary': []}
                value_predictions[value]['true'].extend(y_test[value_mask])
                value_predictions[value]['pred_proba'].extend(y_test_pred_proba[value_mask])
                value_predictions[value]['pred_binary'].extend(y_test_pred_binary[value_mask])
    
    # Calculate overall metrics
    overall_auroc = roc_auc_score(all_true, all_preds)
    overall_accuracy = accuracy_score(all_true, all_preds_binary)
    overall_bal_cal = calculate_balanced_calibration_loss(all_true, all_preds)
    overall_ece = calculate_ece(all_true, all_preds)

    # Calculate per-value metrics
    value_metrics = {}
    for value, preds in value_predictions.items():
        try:
            auroc = roc_auc_score(preds['true'], preds['pred_proba'])
            accuracy = accuracy_score(preds['true'], preds['pred_binary'])
            bal_cal = calculate_balanced_calibration_loss(preds['true'], preds['pred_proba'])
            value_metrics[value] = {
                'AUROC': auroc,
                'Accuracy': accuracy,
                'Balanced Calibration Loss': bal_cal,
                'ECE': overall_ece,
                'Samples': len(preds['true'])
            }
        except ValueError:
            value_metrics[value] = {
                'AUROC': float('nan'),
                'Accuracy': float('nan'),
                'Balanced Calibration Loss': float('nan'),
                'ECE': float('nan'),
                'Samples': len(preds['true'])
            }
    
    print(f"\nOverall Metrics (OOD):")
    print(f"AUROC: {overall_auroc:.3f}")
    print(f"Accuracy: {overall_accuracy:.3f}")
    print(f"Balanced Calibration Loss: {overall_bal_cal:.3f}")
    print(f"ECE: {overall_ece:.3f}")
    print(f"Average training time per fold (OOD): {np.mean(fold_training_times):.4f} seconds")

    prob_true, prob_pred = calibration_curve(all_true, all_preds, n_bins=10)
    
    # Create figure with two subplots
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 12), height_ratios=[2, 1], gridspec_kw={'hspace': 0.1})
    
    # Top plot: Calibration curve
    ax1.plot(prob_pred, prob_true, marker='o', label='Calibration curve')
    ax1.plot([0, 1], [0, 1], linestyle='--', label='Perfect calibration')
    ax1.set_xlabel('')
    ax1.set_ylabel('True Probability')
    ax1.set_title('Calibration Plot')
    ax1.grid(True)
    ax1.legend()
    
    # Bottom plot: Instance percentages
    bin_edges = np.linspace(0, 1, 11)
    bin_counts, _ = np.histogram(all_preds, bins=bin_edges)
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
    bin_percentages = (bin_counts / len(all_preds)) * 100
    
    bars = ax2.bar(bin_centers, bin_percentages, width=0.08, alpha=0.6)
    ax2.set_xlabel('Mean Predicted Probability')
    ax2.set_ylabel('Percentage of Instances (%)')
    ax2.grid(True)
    
    for bar in bars:
        height = bar.get_height()
        ax2.text(bar.get_x() + bar.get_width()/2., height,
                 f'{height:.1f}%',
                 ha='center', va='bottom')
    
    ax1.set_xlim(-0.05, 1.05)
    ax2.set_xlim(-0.05, 1.05)
    
    plt.savefig(f'./figures/calibration_OOD_subject={model_name}_split-by-{split_by}.png', dpi=300, bbox_inches='tight')
    plt.show()
    
    return {
        'overall_auroc': overall_auroc,
        'overall_accuracy': overall_accuracy,
        'overall_balanced_calibration': overall_bal_cal,
        'overall_ece': overall_ece,
        'accumulated_predictions': {
            'y_true': all_true,
            'y_pred_proba': all_preds,
            'y_pred_binary': all_preds_binary
        },
        'value_metrics': value_metrics,
        'fold_training_times': fold_training_times
    }


import os
import glob
import pandas as pd


# List to store results for each subject
results = []

# Iterate over each CSV file
for subject_name in ['babbage-002', 'davinci-002', 'gpt-35-turbo', 'gpt4o', 'o1-mini', 'o1_re=low',
                     'llama3d2-1b', 'llama3d2-3b', 'llama3d2-11b', 'llama3d2-90b', 'llama3d1-405b', 
                     'DK-R1-Dist-Qwen-1.5B', 'DK-R1-Dist-Qwen-7B', 
                     'DK-R1-Dist-Qwen-14B', 'DK-R1-Dist-Qwen-32B']:

    print(subject_name)

    # Directory containing the CSV files
    data_dir = '../ADeLe_battery_data/subject_specific_instance_level_data'
    
    df = pd.read_csv('../ADeLe_battery_data/subject_specific_instance_level_data/' + subject_name + '.csv')

    subject_name = subject_name.replace('DeepSeek-R1-Distill-Qwen', 'DK-R1-Dist-Qwen')
    
    
    # Your existing data processing logic goes here
    X = pd.concat([df['UG'], df.iloc[:, -19:-1]], axis=1)  # level annotations
    X = X.apply(pd.to_numeric, errors='coerce')
    y = df['verification_final']

    # Check for NaN values in X
    nan_rows = X.isna().any(axis=1)
    print(f"Number of rows with NaN values: {nan_rows.sum()}")

    # Remove rows with NaN values from both X and y
    X_clean = X[~nan_rows]
    y_clean = y[~nan_rows]
    df_clean = df[~nan_rows]

    print(f"\nOriginal shape of X: {X.shape}")
    print(f"Shape of X after removing NaN rows: {X_clean.shape}")

    # Verify the cleaned datasets
    print(f"\nAny NaN values remaining in X_clean: {X_clean.isna().any().any()}")

    # Return cleaned versions
    df, X, y = df_clean, X_clean, y_clean
    df = df.reset_index(drop=True)
    X = X.reset_index(drop=True)
    y = y.reset_index(drop=True)
    # Calculate subject accuracy
    subject_accuracy = y.mean()

    # ID evaluation
    results_id = run_ten_fold_ID_evaluation(X, y, model_name=subject_name)


    
    # Run OOD evaluations
    # Example usage for benchmark
    results_benchmark = run_ten_fold_OOD_evaluation(X, y, df, split_by='benchmark', model_name=subject_name)
    
    # Example usage for task
    results_task = run_ten_fold_OOD_evaluation(X, y, df, split_by='task', model_name=subject_name)
    
    # Store results
    result = {
        'subject': subject_name,
        'subject_accuracy': round(subject_accuracy, 3),  # Round to 3 decimals
        'Brier_ID': round(results_id['overall_brier_score'], 3),  # Brier score added for ID data
        'AUROC_ID': round(results_id['overall_auroc'], 3),
        'ECE_ID': round(results_id['overall_ece'], 3),
        'AUROC_task_OOD': round(results_task['overall_auroc'], 3),
        'ECE_task_OOD': round(results_task['overall_ece'], 3),
        'AUROC_benchmark_OOD': round(results_benchmark['overall_auroc'], 3),
        'ECE_benchmark_OOD': round(results_benchmark['overall_ece'], 3)
    }
    results.append(result)

# Create a DataFrame from the results
results_df = pd.DataFrame(results)

# Save to a single CSV file
# results_df.to_csv('predictive_power_stats_rf-assessor.csv', index=False)
```

```python
"""
SAME AS ABOVE BUT SAVING THE INSTANCE-LEVEL RESULTS ON CSVs
"""
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.ensemble import RandomForestClassifier

# Function for instance-wise (ID) evaluation with instance-level predictions
def run_ten_fold_ID_evaluation_with_instance_preds(X, y, df, seed=42):
    kf = KFold(n_splits=10, shuffle=True, random_state=seed)
    preds_list = []
    
    # Loop over folds and record predictions along with the original index
    for train_idx, test_idx in kf.split(X):
        X_train = X.iloc[train_idx]
        X_test = X.iloc[test_idx]
        y_train = y.iloc[train_idx]
        
        model = RandomForestClassifier(random_state=seed, 
                                       min_samples_split=50,
                                       criterion='entropy')
        model.fit(X_train, y_train)
        
        # Get probability for class 1 and binary predictions (threshold 0.5)
        y_pred_proba = model.predict_proba(X_test)[:, 1]
        y_pred_binary = (y_pred_proba > 0.5).astype(int)
        
        # Save predictions along with original indices
        fold_preds = pd.DataFrame({
            'index': X_test.index,
            'class_1_prob': y_pred_proba,
            'y_pred': y_pred_binary
        })
        preds_list.append(fold_preds)
    
    preds_all = pd.concat(preds_list)
    # Merge with original dataframe using the index so that instance_id, question,
    # and verification_final are preserved.
    merged = df.merge(preds_all, left_index=True, right_on='index', how='inner')
    merged = merged[['instance_id', 'question', 'verification_final', 'y_pred', 'class_1_prob']]
    return merged

# Function for OOD evaluation (split by either 'benchmark' or 'task') with instance-level predictions
def run_ten_fold_OOD_evaluation_with_instance_preds(X, y, df, split_by='benchmark', seed=42):
    if split_by not in ['benchmark', 'task']:
        raise ValueError("split_by must be either 'benchmark' or 'task'")
    
    unique_values = df[split_by].unique()
    np.random.seed(seed)
    np.random.shuffle(unique_values)
    
    kf = KFold(n_splits=10, shuffle=True, random_state=seed)
    preds_list = []
    
    # For each fold, determine the train/test split based on the unique values
    for train_val_idx, test_val_idx in kf.split(unique_values):
        train_values = unique_values[train_val_idx]
        test_values = unique_values[test_val_idx]
        
        # Create masks based on the split_by column
        train_mask = df[split_by].isin(train_values)
        test_mask = df[split_by].isin(test_values)
        
        X_train = X.loc[train_mask]
        X_test = X.loc[test_mask]
        y_train = y.loc[train_mask]
        
        model = RandomForestClassifier(random_state=seed, 
                                       min_samples_split=50,
                                       criterion='entropy')
        model.fit(X_train, y_train)
        
        y_pred_proba = model.predict_proba(X_test)[:, 1]
        y_pred_binary = (y_pred_proba > 0.5).astype(int)
        
        fold_preds = pd.DataFrame({
            'index': X_test.index,
            'class_1_prob': y_pred_proba,
            'y_pred': y_pred_binary
        })
        preds_list.append(fold_preds)
    
    preds_all = pd.concat(preds_list)
    merged = df.merge(preds_all, left_index=True, right_on='index', how='inner')
    merged = merged[['instance_id', 'question', 'verification_final', 'y_pred', 'class_1_prob']]
    return merged

# Create output directory if it doesn't exist
output_dir = 'inst_level_rf_predictions'
os.makedirs(output_dir, exist_ok=True)

# List of subjects to process
subjects = ['babbage-002', 'davinci-002', 'gpt-35-turbo', 'gpt4o', 'o1-mini', 'o1_re=low',
            'llama3d2-1b', 'llama3d2-3b', 'llama3d2-11b', 'llama3d2-90b', 'llama3d1-405b', 
            'DK-R1-Dist-Qwen-1.5B', 'DK-R1-Dist-Qwen-7B', 
            'DK-R1-Dist-Qwen-14B', 'DK-R1-Dist-Qwen-32B']

seeds = [1, 2, 3, 4, 5, 42, 123, 456, 789, 1000]

for seed in seeds:
    print(f"\n\nProcessing seed: {seed}")
    for subject in subjects:
        print(f"Processing subject: {subject}")
        # Load the subject CSV from your data directory
        
        subject_file = f"{subject}.csv"
        df = pd.read_csv(os.path.join('../ADeLe_battery_data/subject_specific_instance_level_data/', subject_file))
        
        # (Optional) Adjust subject name if needed
        subject = subject.replace('DeepSeek-R1-Distill-Qwen', 'DK-R1-Dist-Qwen')
        
        # Prepare feature matrix X and label vector y.
        # Here we assume that 'UG' (and some additional columns) form your feature set.
        X = pd.concat([df['UG'], df.iloc[:, -19:-1]], axis=1)
        X = X.apply(pd.to_numeric, errors='coerce')
        y = df['verification_final']
        
        # Remove rows with NaNs in X and adjust the dataframes accordingly
        valid_idx = ~X.isna().any(axis=1)
        X = X[valid_idx].reset_index(drop=True)
        y = y[valid_idx].reset_index(drop=True)
        df = df[valid_idx].reset_index(drop=True)
        
        # --- ID Evaluation ---
        id_preds_df = run_ten_fold_ID_evaluation_with_instance_preds(X, y, df, seed=seed)
        # For ID evaluation, setup string is empty; adjust filename accordingly.
        filename = f"subject_{subject}_combined_seed{seed}.csv"
        id_preds_df.to_csv(os.path.join(output_dir, filename), index=False)
        
        # --- OOD Task Evaluation ---
        ood_task_preds_df = run_ten_fold_OOD_evaluation_with_instance_preds(X, y, df, split_by='task', seed=seed)
        filename = f"subject_{subject}_ood_task_combined_seed{seed}.csv"
        ood_task_preds_df.to_csv(os.path.join(output_dir, filename), index=False)
        
        # --- OOD Benchmark Evaluation ---
        ood_benchmark_preds_df = run_ten_fold_OOD_evaluation_with_instance_preds(X, y, df, split_by='benchmark', seed=seed)
        filename = f"subject_{subject}_ood_benchmark_combined_seed{seed}.csv"
        ood_benchmark_preds_df.to_csv(os.path.join(output_dir, filename), index=False)

        print(f"Saved predictions for subject {subject} into directory '{output_dir}'")


"""
Average the results across 10 seeds
"""
import os
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score

# Helper function to compute Expected Calibration Error (ECE)
def compute_ece(y_true, y_prob, n_bins=10):
    """
    Compute the Expected Calibration Error (ECE) for binary classification.
    y_true: array-like ground truth labels (0 or 1).
    y_prob: array-like predicted probabilities for the positive class.
    n_bins: number of bins to use for calibration error computation.
    """
    bins = np.linspace(0, 1, n_bins + 1)
    bin_indices = np.digitize(y_prob, bins) - 1  # bins start at index 0
    ece = 0.0
    n = len(y_prob)
    
    for i in range(n_bins):
        in_bin = (bin_indices == i)
        count = np.sum(in_bin)
        if count > 0:
            avg_conf = y_prob[in_bin].mean()  # average predicted probability in this bin
            avg_acc = y_true[in_bin].mean()   # average true positive rate in this bin
            ece += np.abs(avg_conf - avg_acc) * (count / n)
    return ece

# Define the list of subjects and their corresponding accuracies (assumed in the same order)
subjects = [
    'babbage-002', 'davinci-002', 'gpt-35-turbo', 'gpt4o', 'o1-mini', 'o1_re=low',
    'llama3d2-1b', 'llama3d2-3b', 'llama3d2-11b', 'llama3d2-90b', 'llama3d1-405b', 
    'DK-R1-Dist-Qwen-1.5B', 'DK-R1-Dist-Qwen-7B', 'DK-R1-Dist-Qwen-14B', 'DK-R1-Dist-Qwen-32B'
]

##### UNLOCK THIS TO CHANGE TO DEEPSEEK ANNOTATIONS #####
# subjects = [s + '_deepseek' for s in subjects]

############################################################
llm_accuracies = [
    0.102,  # babbage-002
    0.157,  # davinci-002
    0.414,  # gpt-35-turbo
    0.713,  # gpt4o
    0.770,  # o1-mini
    0.843,  # o1_re=low (i.e. OpenAI o1)
    0.216,  # llama3d2-1b
    0.378,  # llama3d2-3b
    0.463,  # llama3d2-11b
    0.645,  # llama3d2-90b
    0.683,  # llama3d1-405b
    0.353,  # DK-R1-Dist-Qwen-1.5B
    0.555,  # DK-R1-Dist-Qwen-7B
    0.698,  # DK-R1-Dist-Qwen-14B
    0.748   # DK-R1-Dist-Qwen-32B
]

# Seeds used during evaluation
seeds = [1, 2, 3, 4, 5, 42, 123, 456, 789, 1000]

# Directory where instance-level prediction CSV files are stored
output_dir = 'inst_level_rf_predictions'

# Define evaluation types and the corresponding filename pattern part
eval_types = {
    "ID": "combined",
    "OOD_task": "ood_task_combined",
    "OOD_benchmark": "ood_benchmark_combined"
}

# Initialize a dictionary to store per-seed metrics for each evaluation type and each subject
# Structure: results[eval_type][subject] = {"auroc": [list_of_values], "ece": [list_of_values]}
results = { eval_type: {subject: {"auroc": [], "ece": []} for subject in subjects }
            for eval_type in eval_types }

# Loop over evaluation types, subjects, and seeds to compute per-seed AUROC and ECE
for eval_type, file_suffix in eval_types.items():
    for subject in subjects:
        
        for seed in seeds:
            # Build the filename based on evaluation type, subject name, and seed
            # For ID: subject_{subject}_combined_seed{seed}.csv, etc.
            
            filename = f"subject_{subject}_{file_suffix}_seed{seed}.csv"
            file_path = os.path.join(output_dir, filename)
            
            # Load prediction CSV into a DataFrame
            df = pd.read_csv(file_path)
            
            # Assume 'verification_final' is ground truth and 'class_1_prob' is predicted probability
            y_true = df['verification_final'].values
            y_prob = df['class_1_prob'].values
            
            # Compute AUROC; use try-except in case only one class is present
            try:
                auroc = roc_auc_score(y_true, y_prob)
            except ValueError:
                auroc = np.nan
            # Compute ECE using the helper function
            ece = compute_ece(y_true, y_prob, n_bins=10)
            
            # Append computed metrics to our dictionary
            results[eval_type][subject]["auroc"].append(auroc)
            results[eval_type][subject]["ece"].append(ece)

# Average the metrics for each subject across seeds
subject_metrics = { et: {} for et in eval_types }
for eval_type in eval_types:
    for subject in subjects:
        avg_auroc = np.nanmean(results[eval_type][subject]["auroc"])
        avg_ece = np.nanmean(results[eval_type][subject]["ece"])
        subject_metrics[eval_type][subject] = {"auroc": avg_auroc, "ece": avg_ece}
        print(f"{eval_type} - {subject}: Average AUROC = {avg_auroc:.3f}, Average ECE = {avg_ece:.3f}")

# Compute the weighted average metrics across subjects for each evaluation type.
total_weight = sum(llm_accuracies)
weighted_results = {}

for eval_type in eval_types:
    weighted_auroc = 0.0
    weighted_ece = 0.0
    for subject, acc in zip(subjects, llm_accuracies):
        weighted_auroc += acc * subject_metrics[eval_type][subject]["auroc"]
        weighted_ece += acc * subject_metrics[eval_type][subject]["ece"]
    weighted_auroc /= total_weight
    weighted_ece /= total_weight
    weighted_results[eval_type] = {"weighted_auroc": weighted_auroc, "weighted_ece": weighted_ece}
    print(f"\n{eval_type} - Weighted Metrics across all LLMs:")
    print(f"Weighted AUROC: {weighted_auroc:.3f}")
    print(f"Weighted ECE: {weighted_ece:.3f}")
```

```python
"""
Calculate the Feature Importance
"""
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import KFold

def calculate_feature_importance(X, y, seed=42):
    """
    Calculate and return the feature importance averaged across 10-fold CV.
    """
    kf = KFold(n_splits=10, shuffle=True, random_state=seed)
    all_feature_importances = []

    for train_idx, test_idx in kf.split(X):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

        model = RandomForestClassifier(random_state=seed, min_samples_split=50)
        model.fit(X_train, y_train)
        
        all_feature_importances.append(model.feature_importances_)
    
    feature_importances = np.mean(all_feature_importances, axis=0)
    feature_importances_std = np.std(all_feature_importances, axis=0)
    
    feature_names = X.columns
    return feature_names, feature_importances, feature_importances_std

# Replace this with actual subject names and data directory
subject_names = ['babbage-002', 'davinci-002', 'gpt-35-turbo', 'gpt4o', 'o1-mini', 'o1_re=low',
                 'llama3d2-1b', 'llama3d2-3b', 'llama3d2-11b', 'llama3d2-90b', 'llama3d1-405b', 
                 'DK-R1-Dist-Qwen-1.5B', 'DK-R1-Dist-Qwen-7B', 
                 'DK-R1-Dist-Qwen-14B', 'DK-R1-Dist-Qwen-32B']

# To collect feature importances across subjects
all_subjects_feature_importances = []
all_subjects_feature_names = []

# Iterate over each subject and calculate feature importance
for subject_name in subject_names:
    print(f"Processing feature importance for {subject_name}")
    
    # Load data for the subject (adjust file paths as needed)
    df = pd.read_csv(f'../ADeLe_battery_data/subject_specific_instance_level_data/{subject_name}.csv')

    X = pd.concat([df['UG'], df.iloc[:, -19:-1]], axis=1)  # Extract features
    X = X.apply(pd.to_numeric, errors='coerce')  # Ensure numeric values
    y = df['verification_final']  # Target variable

    nan_rows = X.isna().any(axis=1)
    X_clean = X[~nan_rows]
    y_clean = y[~nan_rows]

    X_clean = X_clean.reset_index(drop=True)
    y_clean = y_clean.reset_index(drop=True)
    
    # Calculate feature importances for each subject
    feature_names, feature_importances, feature_importances_std = calculate_feature_importance(X_clean, y_clean)
    
    all_subjects_feature_importances.append(feature_importances)
    all_subjects_feature_names.append(feature_names)

# Average feature importances across all subjects
average_feature_importance = np.mean(all_subjects_feature_importances, axis=0)
std_feature_importance = np.std(all_subjects_feature_importances, axis=0)

# Create a DataFrame for plotting
feature_importance_df = pd.DataFrame({
    'Feature': all_subjects_feature_names[0],  # Feature names are the same for all subjects
    'Importance': average_feature_importance,
    'Std': std_feature_importance
})

# Sort alphabetically by feature name
feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=True)

# Plot feature importance with error bars (one standard deviation)
plt.figure(figsize=(10, 6))
plt.barh(feature_importance_df['Feature'], feature_importance_df['Importance'], xerr=feature_importance_df['Std'], capsize=5)
plt.xlabel('Feature Importance')
plt.ylabel('Demand')
# plt.title('Feature Importance Averaged Across All Subjects and 10 Folds')
plt.tight_layout()

# Save the plot as a PDF
plt.savefig('./figures/Feature_Importance_Demand-based_Assessor.pdf', format='pdf')

plt.show()
```
